Task 2 — Time Series Analysis & Decomposition

```python
#import Required libraries
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller
```

```python
#create monthly sales data
# Group sales by month
monthly_sales = df.groupby(pd.Grouper(key="Order Date", freq="ME"))["Sales"].sum()
monthly_sales.head()
```

```python
#plot monthly sales trend
plt.figure(figsize=(8,4))
plt.plot(monthly_sales, marker="o")
plt.title("Monthly Sales Trend (2015-2018)")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
#time series decomposition
decomposition = seasonal_decompose(monthly_sales, model="additive", period=12)
```

```python
#plotting all components
fig = decomposition.plot()
fig.set_size_inches(8,4)
plt.show()
```

```python
#trend component
plt.figure(figsize=(12,4))
plt.plot(decomposition.trend)
plt.title("Trend Component")
plt.show()
```

```python
#seasonal component
plt.figure(figsize=(12,4))
plt.plot(decomposition.seasonal)
plt.title("Seasonal Component")
plt.show()
```

```python
#Residual component
plt.figure(figsize=(12,4))
plt.plot(decomposition.resid)
plt.title("Residual Component")
plt.show()
```

```python
# Augmented Dickey-Fuller (ADF) Test
result = adfuller(monthly_sales)
print("ADF Statistic :", result[0])
print("P-value :", result[1])
if result[1] < 0.05:
    print("The data is stationary.")
else:
    print("The data is not stationary.")
```

```python
#applying differencing
monthly_sales_diff = monthly_sales.diff().dropna()
monthly_sales_diff.head()
```

```python
#plotting difference data
plt.figure(figsize=(12,5))
plt.plot(monthly_sales_diff)
plt.title("Monthly Sales After Differencing")
plt.xlabel("Order Date")
plt.ylabel("Differenced Sales")
plt.grid(True)
plt.show()
```

```python
#ADF test after difference
result_diff = adfuller(monthly_sales_diff)
print("ADF Statistic :", result_diff[0])
print("P-value :", result_diff[1])
if result_diff[1] < 0.05:
    print("The differenced data is stationary.")
else:
    print("The differenced data is still not stationary.")
```

#Observations
1. The overall sales trend increases over the four-year period, showing business growth.
2. The seasonal component shows a repeating yearly pattern, with higher sales toward the end of each year.
3. The residual component contains random fluctuations that are not explained by the trend or seasonality.
4. The largest residual variations are typically seen during high-sales periods such as November and December.

Stationarity means that the average, variation, and overall behavior of a time series remain constant over time.
The Augmented Dickey-Fuller (ADF) test is used to check stationarity.

If the p-value is less than 0.05, the series is stationary.

If the p-value is greater than 0.05, the series is non-stationary, so differencing is applied. After differencing, the ADF test is performed again to confirm that the series has become stationary.

```python
#confidence intervals
confidence_interval = forecast.conf_int()
confidence_interval
```

```python
def prepare_segment(data):
    data = data.copy()
    data["Order Date"] = pd.to_datetime(data["Order Date"])
    data = data.set_index("Order Date")
    monthly = (
        data["Sales"]
        .resample("ME")
        .sum()
    )
    return monthly
```

```python
#5 segments
segments = {
    "Furniture": df[df["Category"]=="Furniture"],
    "Technology": df[df["Category"]=="Technology"],
    "Office Supplies": df[df["Category"]=="Office Supplies"],
    "West": df[df["Region"]=="West"],
    "East": df[df["Region"]=="East"]
}
```

```python
#creating forecast function
from xgboost import XGBRegressor
def forecast_segment(data):
    sales = prepare_segment(data)
    temp = sales.to_frame()
    temp.columns = ["Sales"]
    temp["Lag1"] = temp["Sales"].shift(1)
    temp["Lag2"] = temp["Sales"].shift(2)
    temp["Lag3"] = temp["Sales"].shift(3)
    temp["Rolling_Mean"] = (
        temp["Sales"]
        .shift(1)
        .rolling(3)
        .mean()
    )
    temp["Month"] = temp.index.month
    temp["Quarter"] = temp.index.quarter
    temp.dropna(inplace=True)
    X = temp.drop("Sales", axis=1)
    y = temp["Sales"]
    model = XGBRegressor(
        n_estimators=100,
        random_state=42
    )
    model.fit(X,y)
    future = []
    last_data = temp.copy()

    for i in range(3):
        row = last_data.iloc[-1:].drop("Sales",axis=1)
        pred = model.predict(row)[0]
        future.append(pred)
        new_row = row.copy()
        new_row["Lag1"] = pred
    return future
```

```python
#generating forecasts for all segments
segment_forecasts = {}
for name, data in segments.items():
    segment_forecasts[name] = forecast_segment(data)
segment_forecasts
```

```python
#create forecast table
forecast_table = pd.DataFrame(segment_forecasts)
forecast_table.index = [
    "Month 1",
    "Month 2",
    "Month 3"
]
forecast_table
```

```python
#plotting of all 5 forecasts together
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
for column in forecast_table.columns:
    plt.plot(
        forecast_table.index,
        forecast_table[column],
        marker="o",
        label=column
    )
plt.title("Category and Region Sales Forecast Comparison")
plt.xlabel("Forecast Period")
plt.ylabel("Predicted Sales")
plt.legend()
plt.show()
```

**Task 5 — Anomaly Detection in Sales Data**


```python
#importing libraries
from sklearn.ensemble import IsolationForest
```

```python
#preapring weekly sales data
df["Order Date"] = pd.to_datetime(df["Order Date"])
weekly_sales = (
    df.groupby(
        pd.Grouper(
            key="Order Date",
            freq="W"
        )
    )["Sales"]
    .sum()
)
weekly_sales.head()
```

```python
#plotting weekly sales trend
plt.figure(figsize=(12,5))
plt.plot(weekly_sales.index,weekly_sales)
plt.title("Weekly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

**Part 1:Isolation forest**

```python
#preparing data for isolation forest
sales_values = weekly_sales.values.reshape(-1,1)
```

```python
#train isolation forest
iso_model = IsolationForest(contamination=0.05,random_state=42)
weekly_sales_anomaly = iso_model.fit_predict(sales_values)
```

```python
#marking isolation forest anomalies
weekly_result = pd.DataFrame({
    "Sales": weekly_sales,
    "Isolation_Result": weekly_sales_anomaly
})
weekly_result.head()
```

```python
#extracting anomalies
iso_anomalies = weekly_result[weekly_result["Isolation_Result"] == -1]
iso_anomalies
```

```python
#plotting isolation forest anomalies
plt.figure(figsize=(12,5))
plt.plot(
    weekly_result.index,
    weekly_result["Sales"],
    label="Weekly Sales"
)
plt.scatter(
    iso_anomalies.index,
    iso_anomalies["Sales"],
    marker="o",
    label="Anomaly"
)
plt.title("Isolation Forest Sales Anomalies")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

**Part 2: Z-Score Anomaly Detection**

```python
#calculating mean and Z-score
z_data = pd.DataFrame()
z_data["Sales"] = weekly_sales
z_data["Rolling_Mean"] = (
    weekly_sales
    .rolling(4)
    .mean()
)
z_data["Rolling_STD"] = (
    weekly_sales
    .rolling(4)
    .std()
)
z_data["Z_score"] = (
    (z_data["Sales"] - z_data["Rolling_Mean"])/z_data["Rolling_STD"])
z_data.head()
```

```python
#Flag Z-Score Anomalies
z_data["Z_Anomaly"] = (
    abs(z_data["Z_score"]) > 2
)
z_anomalies = z_data[
    z_data["Z_Anomaly"]
]
z_anomalies
```

```python
#plotting Z-Score Anomalies
plt.figure(figsize=(12,5))
plt.plot(
    z_data.index,
    z_data["Sales"],
    label="Weekly Sales"
)
plt.scatter(
    z_anomalies.index,
    z_anomalies["Sales"],
    marker="x",
    label="Z-score Anomaly"
)
plt.title("Z-score Sales Anomalies")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#comparison
iso_dates = set(iso_anomalies.index)
z_dates = set(z_anomalies.index)
print("Isolation Forest anomalies:")
print(iso_dates)
print("\nZ-score anomalies:")
print(z_dates)
print("\nCommon anomalies:")
print(iso_dates.intersection(z_dates))
```

```python
#countr Agreement
same = len(iso_dates.intersection(z_dates))
print(
    "Number of anomalies detected by both methods:",
    same
)
```

```python
iso_anomalies.to_csv("anomaly_results.csv")
print("Anomaly dataset saved!")
```

Conclusion:
Isolation Forest detects anomalies based on unusual patterns compared with the overall dataset, while Z-score detects values that are statistically far from the rolling average. If both methods identify the same weeks, those periods are stronger anomaly candidates. If they disagree, it indicates that one method is capturing different types of unusual behavior: Isolation Forest detects complex patterns, while Z-score only detects extreme numerical deviations.
